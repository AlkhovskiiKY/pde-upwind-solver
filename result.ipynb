# Численное решение уравнение переноса и уравнения Бюргерса

Уравнение переноса:
$$\frac{\partial u}{\partial t}+u\cdot\frac{\partial u}{\partial x} = 0$$
Уравнение Бюргерса:
$$\frac{\partial u}{\partial t}+u\frac{\partial u}{\partial x}=\nu \frac {\partial ^{2}u}{\partial x^{2}}$$

$u-\text{значение нашей функции } u=u(x, t), \ x-\text{координата перемещения}, \ t-\text{время}, \ \nu-\text{вязкость}$.

Уравнение, написанные выше мы и будем моделировать с помощью `C` и использовать `seaborn` и `matplotlib` для постройки графиков.

Для начала в терминале скомпелируйте файлы написанные на языке `C` с помощью команды `make`.

Мы получим файл `result.csv`, содержащий данные по нашему уравнению: `t` - время, `x` - перемещение, `u` - численное решение , `u_exact` - аналитическое решение.

Затем можем запустить код написанный на языке `Python`, в котором используется анализ наших данных с помощью библиотеки `Pandas`, и визуализация основанная на библиотеке `MatPlotLib` и `seaborn`:

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15,10))

df = pd.read_csv("result.csv")
for t, g in df.groupby("t"):
    sns.lineplot(x=g['X'], y=g['U'])
plt.xlabel("x"); plt.ylabel("u", rotation = 0); plt.grid()
plt.title('Уравнение переноса')
plt.savefig('img/equation1.png')
plt.show()
```

На рисунке сверху изображена графически зависимость `u` - численно вычесленного значения от `x` - координаты нашего тела. Так же, как вы можете заметить график получился с ноложением волн - это получается в следствие того, что мы берём все промежутки времени нашего решение и изображаем каждое. Для наглядности изображу решение при фиксированном моменте t:

```python
t_last = df['t'].median()
g = df[df['t'] == t_last]
sns.lineplot(x=g['X'], y=g['U'], label=f'U, t={t_last:.3f}')
plt.xlabel("x"); plt.ylabel("u", rotation = 0); plt.grid()
plt.show()
```

Получим график уравнения Бюргерса:

```python
plt.figure(figsize=(15,10))

df = pd.read_csv("result.csv")
for t, g in df.groupby("t"):
    sns.lineplot(x=g['X'], y=g['U'])
plt.xlabel("x"); plt.ylabel("u", rotation = 0); plt.grid()
plt.title('Уравнение Бюргерса')
plt.savefig('img/equation2.png')
plt.show()
```
